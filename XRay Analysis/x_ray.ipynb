# Part 1 — Setup and data

## 1 · Setup

TensorFlow/Keras builds and trains the CNN. NumPy and Matplotlib are used only to display a batch of training images. `pathlib` is used to find the image folders without hard-coding a single machine path.


```python
import os
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
```

## 2 · Dataset paths

Keras expects **one folder per class** under each split (`train/normal`, `train/pneumonia`, `train/tuberculosis`, and the same under `val` and `test`). The original path was an absolute OneDrive copy of a Kaggle-style `archive/` tree. This cell looks next to the notebook first, then under `XRay Analysis/`, then the original Bellevue path.


```python
# image_dataset_from_directory reads class names from subfolder names, sorted
# alphabetically: normal=0, pneumonia=1, tuberculosis=2.
candidates = [
    Path("archive"),
    Path("XRay Analysis/archive"),
    Path(
        "/Users/michaelperrine/Library/CloudStorage/OneDrive-Personal/"
        "Documents/Bellevue/Applied Data Science/Project 3/archive"
    ),
]
data_root = next((p for p in candidates if (p / "train").is_dir()), None)
if data_root is None:
    raise FileNotFoundError(
        "Could not find archive/train. Place the chest-xray archive next to "
        "this notebook (archive/train, archive/val, archive/test)."
    )

train_dir = str(data_root / "train")
val_dir = str(data_root / "val")
test_dir = str(data_root / "test")
print(f"data_root = {data_root}")
```

## 3 · Class counts

`os.listdir` on each class folder is a cheap imbalance check before any tensor work. Pneumonia is the smallest training class here; tuberculosis is the largest. A model that ignores that skew can look accurate while missing pneumonia — the class a screening tool can least afford to miss.

These counts are **file counts**, including any non-image junk in the folder if it exists. Keras will report the files it actually ingested in section 5.


```python
count_normal = len(os.listdir(os.path.join(train_dir, "normal")))
count_pneumonia = len(os.listdir(os.path.join(train_dir, "pneumonia")))
count_tuberculosis = len(os.listdir(os.path.join(train_dir, "tuberculosis")))
```

```python
print(f"Number of normal images in training set: {count_normal}")
print(f"Number of pneumonia images in training set: {count_pneumonia}")
print(f"Number of tuberculosis images in training set: {count_tuberculosis}")
print(
    "train share — "
    f"normal {count_normal / (count_normal + count_pneumonia + count_tuberculosis):.1%}, "
    f"pneumonia {count_pneumonia / (count_normal + count_pneumonia + count_tuberculosis):.1%}, "
    f"tuberculosis {count_tuberculosis / (count_normal + count_pneumonia + count_tuberculosis):.1%}"
)
```

# Part 2 — Load and inspect

## 4 · Image size and batch

224×224 is a common ImageNet-era square. These films are grayscale, so the extra spatial resolution is paid for in pixels, not in RGB channels. `batch_size=10` is small: 20,450 training images become 2,045 steps per epoch, which makes the loss noisy and training slower than a batch of 32. It is kept here so this baseline matches the original run.


```python
image_height = 224
image_width = 224
batch_size = 10
```

## 5 · Load datasets

`image_dataset_from_directory` infers labels from folder names and yields `(images, labels)` batches. `color_mode='grayscale'` loads a single channel — chest X-rays are not RGB photographs. Images are resized to `(224, 224)` on the way in.

Default shuffle is on for all three calls here. That is fine for train; for a locked test evaluation you would usually pass `shuffle=False` so later confusion-matrix work stays aligned. This notebook only calls `model.evaluate`, so order does not change the reported averages.


```python
train_ds = tf.keras.preprocessing.image_dataset_from_directory(
    train_dir,
    color_mode="grayscale",
    image_size=(image_height, image_width),
    batch_size=batch_size,
)
val_ds = tf.keras.preprocessing.image_dataset_from_directory(
    val_dir,
    color_mode="grayscale",
    image_size=(image_height, image_width),
    batch_size=batch_size,
)
test_ds = tf.keras.preprocessing.image_dataset_from_directory(
    test_dir,
    color_mode="grayscale",
    image_size=(image_height, image_width),
    batch_size=batch_size,
)
```

## 6 · Class names

Integer labels follow this list: index 0 is `normal`, 1 is `pneumonia`, 2 is `tuberculosis`. The final Dense layer has 3 units for that reason. If the folders were named differently, the mapping would change and a saved model would be read wrong.


```python
train_ds.class_names
```

## 7 · Sample images

One training batch is drawn and nine chips are plotted. `np.squeeze` drops the singleton grayscale channel so `imshow` gets a 2D array. This is a sanity check that labels match the films, not a substitute for a radiology review.


```python
plt.figure(figsize=(10, 10))
for images, labels in train_ds.take(1):
    for i in range(9):
        plt.subplot(3, 3, i + 1)
        plt.imshow(np.squeeze(images[i].numpy().astype("uint8")))
        plt.title(train_ds.class_names[labels[i]])
        plt.axis("off")
```

## 8 · Dataset pipeline

`cache()` keeps decoded batches in memory after the first epoch so disk I/O is not repeated. `shuffle(1000)` mixes training order (buffer is smaller than the full set). `prefetch(AUTOTUNE)` overlaps preprocessing with the GPU/CPU train step.

Validation and test are cached and prefetched but **not** shuffled again here. Train is the only split that needs random order for SGD.


```python
autotune = tf.data.AUTOTUNE

train_ds = train_ds.cache().shuffle(1000).prefetch(buffer_size=autotune)
val_ds = val_ds.cache().prefetch(buffer_size=autotune)
test_ds = test_ds.cache().prefetch(buffer_size=autotune)
```

# Part 3 — Convolutional neural network

## 9 · Architecture

A `Sequential` CNN with three identical 32-filter 3×3 ReLU conv blocks, each followed by 2×2 max-pool. That is a small capacity choice: later blocks do not widen (32 → 32 → 32), unlike a typical VGG-style 32 → 64 → 128 stack.

`Rescaling(1./255)` maps pixel values from `[0, 255]` to `[0, 1]` inside the model so the saved `.keras` file includes the same transform at inference.

`Flatten` + `Dense(128)` is a large fully connected head on a 224×224 input after three pools (spatial size 28×28×32 → 25,088 units into 128). There is no dropout in this baseline. The last layer is 3-way **softmax**: class probabilities, not logits.


```python
model = tf.keras.Sequential([
    tf.keras.layers.Rescaling(1.0 / 255),
    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dense(3, activation="softmax"),
])
```

## 10 · Compile

Adam is the optimizer. `SparseCategoricalCrossentropy` matches integer labels (0/1/2) rather than one-hot vectors.

**`from_logits=True` is left as in the original run, and it is the wrong flag for this head.** Softmax already produced probabilities. Keras printed a warning during fit: the loss was told to expect raw logits. That mismatch distorts the training signal and can send reported **loss** to huge values even when **accuracy** still moves. The correct compile for a softmax output is `from_logits=False` (the default). This cell is not silently changed so the baseline stays comparable to `xray model0.keras`; fix it before any new training run.


```python
model.compile(
    optimizer="Adam",
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=["accuracy"],
)
```

## 11 · Train

Ten epochs, validation folder passed as `validation_data` so each epoch reports `val_loss` / `val_accuracy`. There is no `class_weight`, no data augmentation, and no early stopping. Re-running this cell from a fresh kernel retrains from random weights; the saved file on disk is the earlier run.


```python
epochs = 10
```

```python
model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=epochs,
)
```

## 12 · Evaluate

`model.evaluate(test_ds)` is the first time the **test** folder is used. In the original run, test accuracy was about **67.5%** while test loss was thousands — that loss number is not a meaningful cross-entropy when `from_logits=True` is paired with softmax (section 10). Accuracy is still a valid 0–1 hit rate; it is just not a clinical metric, and it can hide a weak pneumonia class.


```python
model.evaluate(test_ds)
```

## 13 · Save

Writes the full Keras model (architecture, weights, and the Rescaling layer) to `xray model0.keras` in the working directory. The filename has a space; load it with the same string. Later experiments in `xray_2.ipynb` write different files (`xray_2.keras`, `best_xray_cnn.keras`).


```python
model.save("xray model0.keras")
```

## 14 · Summary and conclusions

**What was built.** A three-class CNN on grayscale chest X-rays (normal / pneumonia / tuberculosis), trained 10 epochs on a folder-split dataset and scored on a held-out test folder. The same architecture is what `xray model0.keras` stores.

**What the results support**

- Folder labels are the supervision. Keras integer IDs follow alphabetical class names (`normal`, `pneumonia`, `tuberculosis`).
- Grayscale loading matches the modality. Treating films as RGB would invent two empty channels.
- A validation split during `fit` is the right check for this baseline. Train accuracy in the mid-60s to high-60s with similar validation accuracy is **not** a spectacular overfit on accuracy; it is a modest three-way classifier.
- Counting files before training shows the skew: pneumonia is the smallest class, tuberculosis the largest. Majority-class accuracy would be the tuberculosis share (~42% of train), so ~67% test accuracy is better than that dummy rule and still far from a reliable screen.

**What the results do not support**

- **The compile flag is inconsistent.** `from_logits=True` with a softmax output is why Keras warned and why test **loss** exploded. Do not interpret that loss as calibration or as "the model diverged" by itself.
- **Accuracy is not a diagnosis.** There is no confusion matrix, precision, recall, or ROC in this notebook. Missing pneumonia is a different error from missing tuberculosis.
- **No class weights.** The optimizer sees more tuberculosis examples. Pneumonia can be under-predicted without that showing up clearly in a single accuracy number.
- **No augmentation or regularization.** The Dense(128) head on a flattened 28×28×32 map is large; dropout and a wider conv stack appear in `xray_2.ipynb`, not here.
- **This is not a medical device.** Films, labels, and scanners vary. A 67% three-way accuracy on one archive does not license clinical use.

**Extending this.** Set `from_logits=False` (or drop softmax and keep logits). Add `class_weight`, a confusion matrix on `shuffle=False` test data, and early stopping. Those steps, plus augmentation and a pretrained backbone, are the path already started in `xray_2.ipynb`.
